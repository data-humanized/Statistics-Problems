
## Two tailed-TEST

A bakery claims that their cupcakes weigh is 150 grams on average.
A customer suspects this isn't true.
They take a random sample of 10 cupcakes and find:
Sample mean = 145 grams
Sample standard deviation = 8 grams

👉 At a 5% significance level, test the customer's claim using a one-sample T-Test.





*   Ho = 150
*   H1 <> 150  (to make h0 fail)
* So it is a two tailed test




```python
import numpy as np
from scipy import stats
```

```python
mu = 150      # Population mean
x = 145       # Sample mean
n = 10        # Sample Size
df = n - 1    # Degree of freedom
s = 8         # Standard deviation of sample
alpha = 0.05 # Significance Level
```

```python
std_error = s/np.sqrt(n)
t_test = (x - mu) / (s / np.sqrt(n))
t_test
```

```python
p_value = 2 * (1 - stats.t.cdf(abs(t_test), df))
```

```python
p_value
```

```python
if p_value <= alpha:
  print("Reject the Null hypothesis")
else:
  print("Fail to reject the null hypothesis")
```

```python
t_crt_left = stats.t.ppf(alpha/2,df)
t_crit_right = stats.t.ppf(1 - alpha/2,df)

print(f"left margin : {x + t_crt_left * std_error}")
print(f"right margin : {x - t_crt_left * std_error}")
```

## One tailed test
A bakery claims that their cupcakes weigh is atleast 150 grams on average. A customer suspects this isn't true. They take a random sample of 10 cupcakes and find: Sample mean = 145 grams Sample standard deviation = 8 grams

👉 At a 5% significance level, test the customer's claim using a one-sample T-Test.



*   Ho >= 150 (claim by the baker)
*   H1 < 150 (alternate claim)

As there is only one condition to make to Ho fail
It is a One tail test but a left tailed as the condition is "should be smaller"



```python
# o stands for one tail
mu_l = 150      # Population mean
x_l = 145       # Sample mean
n_l = 10        # Sample Size
df_l = n - 1    # Degree of freedom
s_l = 8         # Standard deviation of sample
alpha_l = 0.05 # Significance Level
```

```python
t_score_l = (x_l - mu_l) / (s_l / np.sqrt(n_l))
t_score_l
```

```python
p_value_l = stats.t.cdf(t_score_l,df_l)
p_value_l
```

```python
if p_value_l <= alpha:
  print("Reject the Null hypothesis")
else:
  print("Fail to reject the null hpythesis")
```

```python
t_critic_l = stats.t.ppf(alpha,df_l)
t_critic_l                            # only left margin t_critical is needed   as it is a left tail test
```

```python
if t_critic_l > t_score_l:
  print("Reject the Null Hypothesis ")
else:
  print("Fail to reject the Null Hypothesis")
```
