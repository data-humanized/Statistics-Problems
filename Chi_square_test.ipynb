```python
import numpy as np
from scipy import stats

# 1. Input the observed values from your image (excluding totals)
observed = np.array([[30,10,5],   # Male: Coffee, Tea, Juice
    [15, 20, 20]   # Female: Coffee, Tea, Juice
])

# 2. Set significance level
alpha = 0.05

# 3. Calculate Chi-Square test of independence
chi2_stat, p_val, dof, expected = stats.chi2_contingency(observed, correction=False)

# 4. Calculate critical value
chi2_critical = stats.chi2.ppf(1 - alpha, dof)

# 5. Output results
print(f"Chi-Square Statistic  : {chi2_stat:.4f}")
print(f"Critical Value        : {chi2_critical:.4f}")
print(f"Degrees of Freedom    : {dof}")
print(f"p-value               : {p_val:.6f}")
print("-" * 40)

if chi2_stat > chi2_critical:
    print("CONCLUSION: REJECT the null hypothesis.")
    print("Reason: There is a statistically significant relationship between Gender and Drink preference.")
else:
    print("CONCLUSION: FAIL TO REJECT the null hypothesis.")
    print("Reason: There is no significant relationship between Gender and Drink preference.")
```

```python
import seaborn as sns
import pandas as pd
```

```python
df = sns.load_dataset('titanic')
```

```python
df
```

```python
df.dropna()
```

```python
contingency_table = pd.crosstab(df['sex'],df['survived'])
```

```python
contingency_table
```

```python
chi2, p_value, dof, expected = stats.chi2_contingency(contingency_table)
chi2_critical = stats.chi2.ppf(1-alpha, dof)
```

```python
print(f'Chi-Squared Statistics     :{chi2:.2f}')
print(f'P-Value                    :{p_value:.2f}')
print(f'Degrees of Freedom         :{dof}')
print(f'Chi2-Critical              :{chi2_critical:.2f}')
print(f'Expected Frequencies       :{expected}')
```

```python
if chi2 > chi2_critical:
    print('Reject the null hypothesis')
else:
    print('Fail to reject the null hypothesis')

if p_value < alpha:
    print('Reject the null hypothesis')
else:
    print('Fail to reject the null hypothesis')
```
